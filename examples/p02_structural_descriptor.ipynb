In this tutorial, we will use a pretrained CGCNN model to extract crystal descriptors: the atom features after the convolutions, averaged over each crystal.

Load the `cgcnn_pred` function from the package

```python
import torch

from cgcnn2.pretrained_models import FORMATION_ENERGY_MODEL
from cgcnn2.utils import cgcnn_pred
```

Define the path

```python
data_path = "./data/sample-regression"
model_path = FORMATION_ENERGY_MODEL
```

Inspect the checkpoint

```python
checkpoint = torch.load(model_path, map_location="cpu", weights_only=True)

print(checkpoint.keys())
print(checkpoint["args"])
```

Get the predictions and the descriptors

```python
pred, descriptors = cgcnn_pred(model_path=model_path, full_set=data_path)
```

Print the number of structures

```python
print(len(descriptors))
```

```python
# Print the details of the first descriptor
descriptor = descriptors[0]
print(f"  Type: {type(descriptor)}")
print(f"  Data type: {descriptor.dtype}")
print(f"  Shape: {descriptor.shape}")
print(f"  Size (number of elements): {descriptor.size}")
print(f"  Content:\n{descriptor}\n")
```
